```python
from google.colab import drive
drive.mount('/content/drive')

import numpy as np, pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split

BASE_DIR = Path("/content/drive/MyDrive/Tea/Tea_Quality_3Class")
classes = ["Below Best", "Best", "Poor"]
exts = {".jpg",".jpeg",".png",".webp",".jfif",".bmp"}

rows=[]
for c in classes:
    for p in (BASE_DIR/c).iterdir():
        if p.suffix.lower() in exts:
            rows.append({"path": str(p), "label": c})

df = pd.DataFrame(rows)
print(df["label"].value_counts())

train_df, tmp_df = train_test_split(df, test_size=0.30, stratify=df["label"], random_state=42)
val_df, test_df  = train_test_split(tmp_df, test_size=0.50, stratify=tmp_df["label"], random_state=42)

print("Train:\n", train_df["label"].value_counts())
print("Val:\n",   val_df["label"].value_counts())
print("Test:\n",  test_df["label"].value_counts())
```

```python
import tensorflow as tf
import cv2

IMG_SIZE = (224, 224)
BATCH = 32
label_to_id = {c:i for i,c in enumerate(classes)}

def segment_crop_rgb(path_str: str):
    img = cv2.imread(path_str)  # BGR
    if img is None:
        return np.zeros((*IMG_SIZE,3), dtype=np.float32)

    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    lower = np.array([35, 40, 40], dtype=np.uint8)
    upper = np.array([90, 255, 255], dtype=np.uint8)
    mask = cv2.inRange(hsv, lower, upper)

    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (7,7))
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel, iterations=1)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel, iterations=2)

    # largest component
    n, lab, stats, _ = cv2.connectedComponentsWithStats(mask, connectivity=8)
    if n > 1:
        largest = 1 + np.argmax(stats[1:, cv2.CC_STAT_AREA])
        mask = (lab == largest).astype(np.uint8) * 255
    else:
        mask = np.zeros(mask.shape, dtype=np.uint8)

    # bbox crop
    ys, xs = np.where(mask > 0)
    if len(xs) < 50:
        # fallback: just resize original
        rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        rgb = cv2.resize(rgb, IMG_SIZE)
        return (rgb.astype(np.float32))

    x1, x2 = xs.min(), xs.max()
    y1, y2 = ys.min(), ys.max()

    # pad bbox slightly
    pad = 10
    x1 = max(0, x1 - pad); y1 = max(0, y1 - pad)
    x2 = min(img.shape[1]-1, x2 + pad); y2 = min(img.shape[0]-1, y2 + pad)

    crop = img[y1:y2+1, x1:x2+1]
    crop_mask = mask[y1:y2+1, x1:x2+1]

    crop[crop_mask == 0] = 0  # black background
    rgb = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)
    rgb = cv2.resize(rgb, IMG_SIZE)
    return rgb.astype(np.float32)

def tf_load(path, y):
    img = tf.numpy_function(lambda p: segment_crop_rgb(p.decode("utf-8")), [path], tf.float32)
    img.set_shape((*IMG_SIZE, 3))
    return img, y

def make_ds(df_part, shuffle=True):
    paths = df_part["path"].values.astype(str)
    labels = df_part["label"].map(label_to_id).values.astype(np.int32)
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if shuffle:
        ds = ds.shuffle(1024, seed=42, reshuffle_each_iteration=True)
    ds = ds.map(tf_load, num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(BATCH).prefetch(tf.data.AUTOTUNE)
    return ds

train_ds = make_ds(train_df, shuffle=True)
val_ds   = make_ds(val_df, shuffle=False)
test_ds  = make_ds(test_df, shuffle=False)
```

```python
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.applications.efficientnet import preprocess_input

# Augmentations that directly address your live-demo problem (lighting variation)
augment = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.06),
    layers.RandomZoom(0.10),
    layers.RandomContrast(0.25),
    layers.RandomBrightness(0.20),
], name="augment")

def prep(x, y, training=False):
    if training:
        x = augment(x, training=True)
    x = preprocess_input(x)
    return x, y

train_ds_p = train_ds.map(lambda x,y: prep(x,y,True),  num_parallel_calls=tf.data.AUTOTUNE)
val_ds_p   = val_ds.map(lambda x,y: prep(x,y,False),   num_parallel_calls=tf.data.AUTOTUNE)
test_ds_p  = test_ds.map(lambda x,y: prep(x,y,False),  num_parallel_calls=tf.data.AUTOTUNE)

base = EfficientNetB0(include_top=False, weights="imagenet", input_shape=(*IMG_SIZE,3))
base.trainable = False  # stage 1: train head only

inputs = layers.Input(shape=(*IMG_SIZE,3))
x = base(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.35)(x)
outputs = layers.Dense(3, activation="softmax")(x)
model = Model(inputs, outputs)

# class weights (helps if folder counts differ)
from sklearn.utils.class_weight import compute_class_weight
cw = compute_class_weight(class_weight="balanced", classes=np.array([0,1,2]),
                          y=train_df["label"].map(label_to_id).values)
class_weight = {i: float(cw[i]) for i in range(3)}
print("class_weight:", class_weight)

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

callbacks = [
    tf.keras.callbacks.EarlyStopping(patience=6, restore_best_weights=True, monitor="val_accuracy"),
    tf.keras.callbacks.ReduceLROnPlateau(patience=3, factor=0.5, min_lr=1e-6, monitor="val_loss"),
]

history1 = model.fit(
    train_ds_p,
    validation_data=val_ds_p,
    epochs=20,
    class_weight=class_weight,
    callbacks=callbacks
)

# stage 2: fine-tune top layers
base.trainable = True
for layer in base.layers[:-40]:   # unfreeze only last ~40 layers
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-4),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history2 = model.fit(
    train_ds_p,
    validation_data=val_ds_p,
    epochs=20,
    class_weight=class_weight,
    callbacks=callbacks
)
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix

y_true = np.concatenate([y.numpy() for _, y in test_ds_p], axis=0)
y_prob = model.predict(test_ds_p, verbose=0)
y_pred = np.argmax(y_prob, axis=1)

print("Test report:")
print(classification_report(y_true, y_pred, target_names=classes))
print("Confusion matrix:")
print(confusion_matrix(y_true, y_pred))
```

```python
import numpy as np
import cv2
import tensorflow as tf
import json
import matplotlib.pyplot as plt
from tensorflow.keras.applications.efficientnet import preprocess_input

MODEL_DIR = "/content/drive/MyDrive/Tea/results/finetuned_effnet_leafonly_v1"

model = tf.keras.models.load_model(f"{MODEL_DIR}/model.keras")
meta = json.load(open(f"{MODEL_DIR}/meta.json"))

classes = meta["classes"]
IMG_SIZE = tuple(meta["img_size"])

def segment_crop_rgb(img_bgr):
    hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
    lower = np.array([35, 40, 40], dtype=np.uint8)
    upper = np.array([90, 255, 255], dtype=np.uint8)
    mask = cv2.inRange(hsv, lower, upper)

    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (7,7))
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel, iterations=1)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel, iterations=2)

    # Keep largest component
    n, lab, stats, _ = cv2.connectedComponentsWithStats(mask, connectivity=8)
    if n > 1:
        largest = 1 + np.argmax(stats[1:, cv2.CC_STAT_AREA])
        mask = (lab == largest).astype(np.uint8) * 255
    else:
        mask = np.zeros(mask.shape, dtype=np.uint8)

    ys, xs = np.where(mask > 0)
    if len(xs) < 50:
        rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
        return cv2.resize(rgb, IMG_SIZE), mask

    x1, x2 = xs.min(), xs.max()
    y1, y2 = ys.min(), ys.max()

    pad = 10
    x1 = max(0, x1 - pad); y1 = max(0, y1 - pad)
    x2 = min(img_bgr.shape[1]-1, x2 + pad); y2 = min(img_bgr.shape[0]-1, y2 + pad)

    crop = img_bgr[y1:y2+1, x1:x2+1].copy()
    crop_mask = mask[y1:y2+1, x1:x2+1]
    crop[crop_mask == 0] = 0

    rgb = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)
    rgb = cv2.resize(rgb, IMG_SIZE)
    return rgb, mask

def predict_bgr(img_bgr, show=True):
    rgb, mask = segment_crop_rgb(img_bgr)

    x = preprocess_input(rgb.astype(np.float32))
    x = np.expand_dims(x, 0)

    prob = model.predict(x, verbose=0)[0]
    idx = int(np.argmax(prob))

    pred_label = classes[idx]
    conf = float(prob[idx])

    if show:
        plt.figure(figsize=(10,4))
        plt.subplot(1,2,1)
        plt.title("Original")
        plt.imshow(cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB))
        plt.axis("off")

        plt.subplot(1,2,2)
        plt.title(f"Leaf crop → {pred_label} ({conf*100:.1f}%)")
        plt.imshow(rgb)
        plt.axis("off")
        plt.show()

        print("Probabilities:")
        for c, p in zip(classes, prob):
            print(f"  {c:12s}: {p*100:5.1f}%")

    return pred_label, conf, prob
```
